```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 1. Load the data (using tab separator since your data is spaced by tabs)
# We use error_bad_lines=False or on_bad_lines='skip' just in case there are formatting quirks
df = pd.read_csv('/Users/cmbreen/Documents/snowpoles/snowpoles_paper/model_outputs/rfdetr_200e_bs15_130images/metrics.csv')

# 2. Clean the data (some rows have empty values for validation/training loss)
# We will create two separate clean tables for training and validation plotting
train_data = df.dropna(subset=['train/loss'])
val_data = df.dropna(subset=['val/loss'])

# 3. Create the plot
plt.figure(figsize=(10, 6))

# Plot Training Loss
plt.plot(train_data['step'], train_data['train/loss'], 
         label='Train Loss', color='blue', marker='o', markersize=4, linestyle='-')

# Plot Validation Loss
plt.plot(val_data['step'], val_data['val/loss'], 
         label='Validation Loss', color='orange', marker='s', markersize=6, linestyle='--')

# 4. Add labels and styling
plt.title('Training and Validation Loss over Steps', fontsize=14)
plt.xlabel('Training Steps', fontsize=12)
plt.ylabel('Total Loss', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend(fontsize=12)

# 5. Show the plot!
plt.tight_layout()
plt.show()

# Optional: Save the plot as an image
# plt.savefig('loss_plot.png', dpi=300)
```
